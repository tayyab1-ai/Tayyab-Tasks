```python
"""
Task 8 — Program to find Area of Triangle
"""
a = float(input("Enter length of side a: "))
b = float(input("Enter length of side b: "))
c = float(input("Enter length of side c: "))
s = (a + b + c) / 2  # semi-perimeter
area = (s * (s - a) * (s - b) * (s - c)) ** 0.5
print(f"Area of triangle with sides {a}, {b}, {c} is {area}")
```

```python
"""
Task 13 — Nested IF-ELSE
"""
i=10
if (i==10):
    if (i < 15):
        print("i is less than 15")
    if (i < 12):
        print("i is less than 12 too")
    else:
        print("i is greater than 15")
```

```python
"""
Task 15 — For loops (GeeksforGeeks)
Problem: Print squares of numbers 1..5 and sum a list.
Topic: https://www.geeksforgeeks.org/python-for-loops/
"""
total = 0
for n in range(1, 6):
    print(n, n**2)
for item in [10, 20, 30]:
    total += item
print('sum', total)
```

```python
"""
Task 16 — While loops
Problem: Countdown from 5 to 1 using while.
Topic: https://www.geeksforgeeks.org/python-while-loops/
"""
n = 5
while n >= 1:
    print(n)
    n -= 1
print('Go!')
```

```python
"""
Task 17 — Break statement
Problem: Find first number divisible by 7 in a list.
Topic: https://www.geeksforgeeks.org/python-break-statement/
"""
nums = [3, 11, 14, 21, 8]
for x in nums:
    if x % 7 == 0:
        print('first divisible by 7:', x)
        break
```

```python
"""
Task 18 — Continue statement
Problem: Print only odd numbers from 1 to 10.
Topic: https://www.geeksforgeeks.org/python-continue-statement/
"""
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i)
```

```python
"""
Task 19 — Looping techniques (enumerate, zip)
Problem: Use enumerate and zip on names and scores.
Topic: https://www.geeksforgeeks.org/looping-techniques-python/
"""
names = ['Sara', 'Omar', 'Lina']
scores = [90, 85, 92]
for i, name in enumerate(names):
    print(i, name, scores[i])
for name, score in zip(names, scores):
    print(name, '->', score)
```

```python
"""
Task 21 — Lists (Programiz topic)
Problem: List CRUD: append, insert, pop, slice.
Topic: https://www.programiz.com/python-programming/list
"""
fruits = ['apple', 'banana']
fruits.append('cherry')
fruits.insert(1, 'mango')
removed = fruits.pop()
print(fruits, 'removed', removed)
print(fruits[::-1])
```

```python
"""
Task 22 — Tuples
Problem: Tuple packing, unpacking, immutability.
Topic: https://www.programiz.com/python-programming/tuple
"""
point = (3, 4)
x, y = point
try:
    point[0] = 9
except TypeError as e:
    print('Immutable:', e)
print(x + y)
```

```python
"""
Task 23 — Strings
Problem: Split, join, strip, replace.
Topic: https://www.programiz.com/python-programming/string
"""
line = '  ai, ml, dl  '
parts = [p.strip() for p in line.split(',')]
print('-'.join(parts).upper())
```

```python
"""
Task 24 — Sets
Problem: Unique tags and set operations.
Topic: https://www.programiz.com/python-programming/set
"""
tags = set('data data science ai'.split())
tags.add('ml')
print(sorted(tags))
print('ai' in tags)
```

```python
"""
Task 25 — Dictionaries
Problem: Word frequency counter.
Topic: https://www.programiz.com/python-programming/dictionary
"""
text = 'ai ml ai dl ml ai'
freq = {}
for w in text.split():
    freq[w] = freq.get(w, 0) + 1
print(freq)
```
