```python
"""
Task 41 — Pandas: Pandas basics
Problem: Create Series and DataFrame.
Topic: https://www.w3schools.com/python/pandas_tutorial.asp
"""
import pandas as pd
s = pd.Series([10,20,30], name='values')
df = pd.DataFrame({'A':[1,2],'B':[3,4]})
print(s)
print(df)
```

```python
"""
Task 42 — Pandas: Getting started
Problem: Check pandas version and simple DataFrame.
Topic: https://www.w3schools.com/python/pandas_getting_started.asp
"""
import pandas as pd
print(pd.__version__)
df = pd.DataFrame({'city':['Lahore','Karachi'],'pop_m':[11,16]})
print(df.info())
```

```python
"""
Task 43 — Pandas: Series
Problem: Series indexing and attributes.
Topic: https://www.w3schools.com/python/pandas_series.asp
"""
import pandas as pd
s = pd.Series([5,10,15], index=['a','b','c'])
print(s['b'], s.values, s.index.tolist())
```

```python
"""
Task 44 — Pandas: DataFrames
Problem: Select columns and rows.
Topic: https://www.w3schools.com/python/pandas_dataframes.asp
"""
import pandas as pd
df = pd.DataFrame({'name':['A','B'],'score':[80,90]})
print(df[['name']])
print(df.iloc[1])
```

```python
"""
Task 45 — Pandas: CSV
Problem: Write and read CSV in memory.
Topic: https://www.w3schools.com/python/pandas_csv.asp
"""
import pandas as pd
from io import StringIO
df = pd.DataFrame({'x':[1,2],'y':[3,4]})
csv = df.to_csv(index=False)
loaded = pd.read_csv(StringIO(csv))
print(loaded)
```

```python
"""
Task 46 — Pandas: Analyzing data
Problem: describe() and mean.
Topic: https://www.w3schools.com/python/pandas_analyzing.asp
"""
import pandas as pd
df = pd.DataFrame({'age':[20,22,21,23]})
print(df.describe())
print('mean age', df['age'].mean())
```

```python
"""
Task 47 — Pandas: Cleaning
Problem: Drop duplicates and fill missing values.
Topic: https://www.w3schools.com/python/pandas_cleaning.asp
"""
import pandas as pd
import numpy as np
df = pd.DataFrame({'v':[1,np.nan,1,2]})
df = df.drop_duplicates().fillna(0)
print(df)
```

```python
"""
Task 48 — Pandas: Correlation
Problem: Correlation matrix on numeric columns.
Topic: https://www.w3schools.com/python/pandas_correlations.asp
"""
import pandas as pd
df = pd.DataFrame({'study_hrs':[1,2,3,4],'score':[50,60,70,80]})
print(df.corr(numeric_only=True))
```
