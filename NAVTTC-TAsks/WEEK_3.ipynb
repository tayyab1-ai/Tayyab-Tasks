```python
"""
Task 28 — NumPy: Creating arrays
Problem: Create 1D, 2D, and 3D arrays.
Topic: https://www.w3schools.com/python/numpy_creating_arrays.asp
"""
import numpy as np
a = np.array([1,2,3])
b = np.array([[1,2],[3,4]])
c = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(a.ndim, b.shape, c.size)
```

```python
"""
Task 29 — NumPy: Indexing
Problem: Index and access elements in 2D array.
Topic: https://www.w3schools.com/python/numpy_array_indexing.asp
"""
import numpy as np
arr = np.array([[10,20,30],[40,50,60]])
print(arr[1,2], arr[0], arr[1,:])
```

```python
"""
Task 30 — NumPy: Slicing
Problem: Slice rows and columns.
Topic: https://www.w3schools.com/python/numpy_array_slicing.asp
"""
import numpy as np
arr = np.arange(12).reshape(3,4)
print(arr[1:3, 1:3])
```

```python
"""
Task 31 — NumPy: Data types
Problem: Create arrays with int32 and float64.
Topic: https://www.w3schools.com/python/numpy_data_types.asp
"""
import numpy as np
i = np.array([1,2,3], dtype=np.int32)
f = np.array([1.1,2.2], dtype=np.float64)
print(i.dtype, f.dtype)
```

```python
"""
Task 32 — NumPy: Copy vs view
Problem: Show that slice is a view.
Topic: https://www.w3schools.com/python/numpy_copy_vs_view.asp
"""
import numpy as np
x = np.array([1,2,3,4])
y = x[1:3]
y[0] = 99
print('x', x, 'y', y)
```

```python
"""
Task 33 — NumPy: Shape
Problem: Read and interpret .shape.
Topic: https://www.w3schools.com/python/numpy_array_shape.asp
"""
import numpy as np
m = np.zeros((2,3))
print(m.shape, m.size, m.ndim)
```

```python
"""
Task 34 — NumPy: Reshape
Problem: Reshape 1D to 2D.
Topic: https://www.w3schools.com/python/numpy_array_reshape.asp
"""
import numpy as np
v = np.arange(6)
m = v.reshape(2,3)
print(v, '\n', m)
```

```python
"""
Task 35 — NumPy: Iterating
Problem: Iterate rows of 2D array.
Topic: https://www.w3schools.com/python/numpy_array_iterating.asp
"""
import numpy as np
arr = np.array([[1,2],[3,4]])
for row in arr:
    print(row)
```

```python
"""
Task 36 — NumPy: Join arrays
Problem: Stack with vstack/hstack.
Topic: https://www.w3schools.com/python/numpy_array_join.asp
"""
import numpy as np
a = np.array([1,2,3])
b = np.array([4,5,6])
print(np.vstack((a,b)))
print(np.hstack((a,b)))
```

```python
"""
Task 37 — NumPy: Split arrays
Problem: Split array into equal parts.
Topic: https://www.w3schools.com/python/numpy_array_split.asp
"""
import numpy as np
arr = np.arange(9)
parts = np.array_split(arr, 3)
print(parts)
```

```python
"""
Task 38 — NumPy: Search
Problem: Find index where value equals 50.
Topic: https://www.w3schools.com/python/numpy_array_search.asp
"""
import numpy as np
arr = np.array([10,20,50,40])
print(np.where(arr == 50)[0][0])
```

```python
"""
Task 39 — NumPy: Sort
Problem: Sort ascending.
Topic: https://www.w3schools.com/python/numpy_array_sort.asp
"""
import numpy as np
arr = np.array([3,1,4,1,5])
print(np.sort(arr))
```

```python
"""
Task 40 — NumPy: Random
Problem: Sample random integers and floats.
Topic: https://www.w3schools.com/python/numpy_random.asp
"""
import numpy as np
np.random.seed(0)
print(np.random.randint(1, 10, size=5))
print(np.random.rand(3))
```
