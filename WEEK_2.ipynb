```python
"""
Task 20 — User-defined functions
Problem: Write functions with default args and return multiple values.
Topic: https://www.geeksforgeeks.org/functions-in-python/
"""
def rectangle_area(length, width=1):
    return length * width, 2 * (length + width)

a, p = rectangle_area(5, 3)
print('area', a, 'perimeter', p)
```

```python
"""
Task 26 — Exception handling
Problem: Safe division with try/except.
Topic: https://www.programiz.com/python-programming/exception-handling
"""
def safe_div(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return None
    finally:
        print('attempt finished')

print(safe_div(10, 2), safe_div(10, 0))
```

```python
"""
Task 27 — User-defined exceptions
Problem: Validate age with custom exception.
Topic: https://www.programiz.com/python-programming/user-defined-exception
"""
class InvalidAgeError(Exception):
    pass

def set_age(age):
    if age < 0 or age > 120:
        raise InvalidAgeError('age out of range')
    return age

try:
    set_age(150)
except InvalidAgeError as e:
    print(e)
```

```python
"""
Task 49 — Mean, median, mode
Problem: Compute central tendency for a small dataset.
Topic: https://www.w3schools.com/python/python_ml_mean_median_mode.asp
"""
from statistics import mean, median, mode

data = [2, 3, 3, 5, 7, 11]
print('mean', mean(data))
print('median', median(data))
print('mode', mode(data))
```

```python
"""
Task 50 — Standard deviation
Problem: Variance and standard deviation.
Topic: https://www.w3schools.com/python/python_ml_standard_deviation.asp
"""
import statistics as stats
data = [10, 12, 14, 18, 22]
print('stdev', round(stats.stdev(data), 2))
print('pvariance', round(stats.pvariance(data), 2))
```
